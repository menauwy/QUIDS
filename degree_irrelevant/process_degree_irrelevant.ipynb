# generate random degree data

```python
import json
import random
test_origianl = '/home/ywang1/data1/dataset/q2id/Q2ID_data/test_data_original.jsonl'
test_10 = '/home/ywang1/data1/dataset/q2id/Q2ID_data/test_data_v10.jsonl'
output_file = f'/home/ywang1/workspace/IntentRewriter/src/degree_irrelevant/random_degree/random_degree.jsonl'
with open(test_10, 'r') as f:
    test_10_data = [json.loads(line) for line in f]
# for data in test_origianl, random select on neg data, and retore in  the output file
with open(test_origianl, 'r') as f:
    test_data = [json.loads(line) for line in f]

for data in test_data:
    sample_id = data['sample_id']
    negs = data['neg']
    if len(negs) > 1:
        random.shuffle(negs)
        data['neg'] = [negs[0]]
    else:
        data['neg'] = ['']
    # get ranked pos from test_10_data
    for data_10 in test_10_data:
        if data_10['sample_id'] == sample_id:
            data['pos'] = data_10['pos']
            break

    
with open(output_file, 'w') as f:
    for data in test_data:
        f.write(json.dumps(data) + '\n')
    
```

```python
import json
# for each degree (low_augment, low_empty, random_degree, without_degree) folder, print average rouge score. rouge scores files are rouge.json, rouge_2.json, rouge_4.json, rouge_5.json in the folder
for degree in ['low_empty', 'random_degree', 'without_degree']: #'low_augment', 
    dir_path = f'/home/ywang1/workspace/IntentRewriter/src/degree_irrelevant/{degree}'
    rouge_path = f'{dir_path}/rouge.json'
    rouge_2_path = f'{dir_path}/rouge_2.json'
    rouge_4_path = f'{dir_path}/rouge_4.json'
    rouge_5_path = f'{dir_path}/rouge_5.json'
    rouge_paths = [rouge_path, rouge_2_path, rouge_4_path, rouge_5_path]

    # Initialize variables to store sum of scores
    sum_rouge1 = sum_rouge2 = sum_rougeL = sum_bert = 0
    count = 0

    # Print rouge scores and calculate sum
    for rouge_path in rouge_paths:
        with open(rouge_path, 'r') as f:
            rouge = json.load(f)
            rouge1 = rouge['rouge3']['rouge1_recall']
            rouge2 = rouge['rouge3']['rouge2_recall']
            rougeL = rouge['rouge3']['rougeL_recall']
            bert = rouge['bert']['recall']
            print(f'Rouge scores for {rouge_path}:')
            print(f'ROUGE-1: {rouge1:.4f}, ROUGE-2: {rouge2:.4f}, ROUGE-L: {rougeL:.4f}, BERT: {bert:.4f}')
            
            sum_rouge1 += rouge1
            sum_rouge2 += rouge2
            sum_rougeL += rougeL
            sum_bert += bert
            count += 1

    # Calculate and print average scores
    avg_rouge1 = sum_rouge1 / count
    avg_rouge2 = sum_rouge2 / count
    avg_rougeL = sum_rougeL / count
    avg_bert = sum_bert / count
    print(f'Average Rouge scores for {degree}:')
    print(f'ROUGE-1: {avg_rouge1:.4f}, ROUGE-2: {avg_rouge2:.4f}, ROUGE-L: {avg_rougeL:.4f}, BERT: {avg_bert:.4f} \n')
    
```
