```python
# Import seaborn
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib as mpl
import matplotlib

# 准备数据
# data = {'Method': ['FedAvg', 'FedAvg', 'dynamic parameter aggregation', 'dynamic parameter aggregation', 'dynamic learning speed', 'dynamic learning speed', 'PRDO', 'PRDO'],
#         'Backend model': ['BigRec-based', 'RecFormer-based', 'BigRec-based', 'RecFormer-based', 'BigRec-based', 'RecFormer-based', 'BigRec-based', 'RecFormer-based'],
#         'NDCG@20': [0.0674, 0.073, 0.0675, 0.073, 0.0796, 0.0766, 0.0798, 0.0777]}
data = {'Method': ['FedAvg', 'FedAvg', 'Dynamic Agg', 'Dynamic Agg', 'Dynamic Speed', 'Dynamic Speed', 'FELLRec', 'FELLRec'],
        'Backend model': ['BigRec-based', 'RecFormer-based', 'BigRec-based', 'RecFormer-based', 'BigRec-based', 'RecFormer-based', 'BigRec-based', 'RecFormer-based'],
        'NDCG@10': [0.0126, 0.0124, 0.0139, 0.0166, 0.0137, 0.0169, 0.0141, 0.0170]}
df = pd.DataFrame(data)
plt.figure(figsize=(8, 6))


# 绘制分组柱状图
custom_palette = sns.color_palette(["#EDC6B1", "#B7B7B7", "#7C96AB", "#BFCCB5"])

sns.barplot(x="Backend model", y="NDCG@10", hue="Method", data=df, width=0.5, palette=custom_palette)
plt.ylim(0.011, 0.0178)
sns.set(style="darkgrid")
matplotlib.rcParams.update({
    "font.family": 'Times New Roman',
    "font.size": 25,
})
plt.legend(loc='upper left', fontsize=16)
legend = plt.gca().get_legend()
for text in legend.get_texts():
    text.set_fontweight('bold')
ax = plt.gca()  # 获取当前的Axes对象ax
ax.set_xticklabels(ax.get_xticklabels(), fontsize=24, fontweight='bold')  # 首先设置所有标签以确保都被访问
ax.set_xlabel('Backend Model', fontweight='bold', fontsize=24)
ax.set_ylabel('NDCG@20', fontweight='bold', fontsize=24)
# 增大轴刻度的字体大小
ax.tick_params(axis='both', labelsize=24) # 对ax1的x轴和y轴刻度标签字体大小进行调整
for label in ax.get_xticklabels():
    label.set_fontweight('bold')
plt.title("Games", fontweight='bold', fontsize=24)
matplotlib.rcParams.update({
    "font.family": 'Times New Roman',
    "font.size": 25,
})
plt.tight_layout()
plt.savefig("abl_1.pdf")

plt.show()
```
