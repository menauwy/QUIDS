```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import matplotlib
# 生成示例数据
x = [5,10,20,50,100]  # 100个训练周期

# 生成三组数据，模拟不同条件下的loss收敛情况
y1 = [0.0262,0.0236,0.0183,0.0162,0.0133]
y2 = [0.0373,0.0285,0.0248, 0.0211,0.0195]
fig, ax1 = plt.subplots()
# First y-axis
color = '#F2A154'
ax1.set_xlabel('')
ax1.tick_params(axis='both', labelsize=17) 
# ax1.set_ylabel('FedAvg', color=color, fontweight='bold', fontsize=18)
ax1.set_ylim(0.01, 0.04) 
ax1.plot(x, y1, color=color, label='FedAvg', linewidth=2.5, marker='o')
ax1.tick_params(axis='y', labelcolor='black', labelsize='large')
ax1.set_xticks(x)  # Set x-ticks to match x values

# Create a twin Axes sharing the xaxis
# ax2 = ax1.twinx()  # Second y-axis
color = '#174863'
# ax2.set_ylabel('FELLRec', color=color, fontweight='bold', fontsize=18)
# ax2.set_ylim(0.002, 0.015) 
ax1.plot(x, y2, color=color, label='FELLRec', linewidth=2.5, marker='o')
# ax2.tick_params(axis='y', labelcolor=color, labelsize='large')

# Legend
fig.legend(loc="upper right", bbox_to_anchor=(0.9,0.9), prop={'size': 16, 'weight': 'bold'})

# set y axis label
plt.ylabel("Recall@20", fontsize=18, fontweight='bold')
# set x axis label
plt.xlabel("Client Number", fontsize=18, fontweight='bold')

sns.set(style="darkgrid")
matplotlib.rcParams.update({
    "font.family": 'Times New Roman',
    "font.size": 15,
})
# plt.figure(figsize=(4, 4))
plt.tight_layout()
plt.savefig("client_1.pdf")
plt.show()
```
