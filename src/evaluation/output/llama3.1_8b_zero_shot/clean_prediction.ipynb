```python
import os
import json
import re

run = 1
file_path = f'/home/ywang1/workspace/IntentRewriter/src/evaluation/output/llama3.1_8b_zero_shot/predictions_{run}.jsonl'
output_file = f'clean_prediction_{run}.jsonl'
data = []

with open(file_path, 'r') as file:
    for line in file:
        data.append(json.loads(line))

def find_pattern(text):
    # Regular expression pattern to find text between :\n\n" and "\n\n
    pattern = r':\n\n"(.*?)"\n\n'

    # Find all matches
    matches = re.findall(pattern, text)

    return matches

count = 0
with open(output_file, 'w') as file:
    for i, item in enumerate(data):
        # if i == 1:
        sample_id = item['sample_id']
        prediction = item['intent'][0]['content']
        matches = find_pattern(prediction)
        if len(matches) > 1:
            print(sample_id, matches)
        if len(matches) > 0:
            count += 1
            item['intent'][0]['content'] = matches[0]
            file.write(json.dumps(item) + '\n')
```

# get average rouge score

```python
import os
import json

r_1 = []
r_2 = []
r_l = []
b_r = []
for i in range(1, 6):
    dir_path = f'/home/ywang1/workspace/IntentRewriter/src/evaluation/output/llama3.1_8b_zero_shot/rouge_{i}.json'
    with open(dir_path, 'r') as file:
        data = json.load(file)
        r_1.append(data['rouge3']['rouge1_recall'])
        r_2.append(data['rouge3']['rouge2_recall'])
        r_l.append(data['rouge3']['rougeL_recall'])
        b_r.append(data['bert']['recall'])

print(f'rouge1_fmeasure: {sum(r_1) / len(r_1)}')
print(f'rouge2_fmeasure: {sum(r_2) / len(r_2)}')
print(f'rougeL_fmeasure: {sum(r_l) / len(r_l)}')
print(f'bert_recall: {sum(b_r) / len(b_r)}')    
```
