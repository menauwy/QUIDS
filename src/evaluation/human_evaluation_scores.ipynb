# postprocessing gpt4-0 results

```python
import re

def extract_fluency_scores(strings, instruction):
    scores = []
    for string in strings:
        # Normalize the string by making it lowercase
        normalized_string = string.lower()
        
        # Regular expression to match "fluency score: " followed by a number 1-
        pattern = r'{} score:\s*([1-5])'.format(instruction)
        match = re.search(pattern, normalized_string)
        
        # If a match is found, extract the score
        if match:
            scores.append(int(match.group(1)))  # Convert the score to an integer
        else:
            # Try to capture standalone numbers between 1 and 5 if "fluency score" is not present
            standalone_match = re.match(r'^\s*([1-5])\s*$', string)
            if standalone_match:
                scores.append(int(standalone_match.group(1)))
    
    return scores
```

```python
import json
model = 'mybartlarge' #mybartlarge
instruction = 'exclusion' # inclusion
input_path = f'/home/ywang1/workspace/IntentRewriter/src/evaluation/output/{instruction}/normalized_gpt-4o_{model}_{instruction} score.jsonl'
outut_path = f'/home/ywang1/workspace/IntentRewriter/src/evaluation/output/{instruction}/processed_gpt-4o_{model}_{instruction} score.jsonl'

new_data = []
# read the input jsonl file
with open(input_path, 'r') as f:
    for line in f:
        data = json.loads(line)
        key = list(data.keys())[0]
        strings = list(data.values())[0]
        scores = extract_fluency_scores(strings, instruction)
        if len(scores) == 0:
            print(f'No fluency scores found for sample {key}')
            print(key)
            mean = 0
        else:
            mean = sum(scores) / len(scores)
        new_dict = {'sample_id': key, 'mean': mean, 'scores': scores}
        new_data.append(new_dict)

# write the output jsonl file
with open(outut_path, 'w') as f:
    for item in new_data:
        f.write(json.dumps(item) + '\n')
```

```python
def get_freq_matrix(scores):
    freq_matrix, cat = aggregate_raters(scores, n_cat=6)
    # print(cat)
    if np.any(freq_matrix[:,0] > 0):
        print(f'{scores} has 0s')
    # new_matrix = freq_matrix[:,1:]
    new_matrix = freq_matrix
    print(new_matrix.shape)
    return new_matrix

fluent_our_freq_matrix = get_freq_matrix(fluent_our)
factual_our_freq_matrix = get_freq_matrix(factual_our)
inclusion_our_freq_matrix = get_freq_matrix(inclusion_our)
exclusion_our_freq_matrix = get_freq_matrix(exclusion_our)

fluent_sota_freq_matrix = get_freq_matrix(fluent_sota)
factual_sota_freq_matrix = get_freq_matrix(factual_sota)
inclusion_sota_freq_matrix = get_freq_matrix(inclusion_sota)
exclusion_sota_freq_matrix = get_freq_matrix(exclusion_sota)

```
