```python
import json
test_path = '/home/ywang1/workspace/IntentRewriter/src/intent_type/test_data_v10.jsonl'
test_data = []
test_ids = []
with open(test_path, 'r') as f:
    for line in f:
        test_data.append(json.loads(line))
        test_ids.append(json.loads(line)['sample_id'])
ddl_ids = [58, 4, 26, 94, 53, 159, 172, 217, 204] # 9 
robust_ids = [414, 443, 257, 474, 385, 262, 458, 375, 355, 244, 260] # 11
# the rest ids are semeval
semeval_ids = []
for id in test_ids:
    if id not in ddl_ids and id not in robust_ids:
        semeval_ids.append(id)
len(semeval_ids)
```

```python
ddl_path = '/home/ywang1/workspace/IntentRewriter/src/intent_type/test_ddl_v10.jsonl'
robust_path = '/home/ywang1/workspace/IntentRewriter/src/intent_type/test_robust_v10.jsonl'
semeval_path = '/home/ywang1/workspace/IntentRewriter/src/intent_type/test_semeval_v10.jsonl'

# Split test data into 3 files according to the sample_id
with open(ddl_path, 'w') as ddl_file, open(robust_path, 'w') as robust_file, open(semeval_path, 'w') as semeval_file:
    for item in test_data:
        if item['sample_id'] in ddl_ids:
            json.dump(item, ddl_file)
            ddl_file.write('\n')
        elif item['sample_id'] in robust_ids:
            json.dump(item, robust_file)
            robust_file.write('\n')
        else:
            json.dump(item, semeval_file)
            semeval_file.write('\n')
```

```python
# Initialize dictionaries to store sample_id and intent for each data source
ddl_data = []
robust_data = []
semeval_data = []

# Function to read data from a file and populate the corresponding dictionary
def populate_data(file_path, data_list):
    with open(file_path, 'r') as f:
        for line in f:
            item = json.loads(line)
            data_list.append({
                'sample_id': item['sample_id'],
                'intent': item['intent']
            })
# Populate data for each source
populate_data(ddl_path, ddl_data)
populate_data(robust_path, robust_data)
populate_data(semeval_path, semeval_data)

# Print the number of items in each dictionary to verify
print(f"DDL data: {len(ddl_data)} items")
print(f"ROBUST data: {len(robust_data)} items")
print(f"SemEval data: {len(semeval_data)} items")

# Optionally, print a sample from each dictionary
print("\nSample from DDL:")
print(ddl_data[0])
print("\nSample from ROBUST:")
print(robust_data[0])
print("\nSample from SemEval:")
print(semeval_data[0])
```
