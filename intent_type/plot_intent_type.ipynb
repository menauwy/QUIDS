```python
# split all data into 2 groups: the first 9 are expl, the rest are info
expl_fluent_our_human = fluent_our_human[:9]
info_fluent_our_human = fluent_our_human[9:]
expl_factual_our_human = factual_our_human[:9]
info_factual_our_human = factual_our_human[9:]
expl_inclusion_our_human = inclusion_our_human[:9]
info_inclusion_our_human = inclusion_our_human[9:]
expl_exclusion_our_human = exclusion_our_human[:9]
info_exclusion_our_human = exclusion_our_human[9:]

expl_fluent_our_gpt = fluent_our_gpt[:9]
info_fluent_our_gpt = fluent_our_gpt[9:]
expl_factual_our_gpt = factual_our_gpt[:9]
info_factual_our_gpt = factual_our_gpt[9:]
expl_inclusion_our_gpt = inclusion_our_gpt[:9]
info_inclusion_our_gpt = inclusion_our_gpt[9:]
expl_exclusion_our_gpt = exclusion_our_gpt[:9]
info_exclusion_our_gpt = exclusion_our_gpt[9:]

expl_fluent_our_llama = fluent_our_llama[:9]
info_fluent_our_llama = fluent_our_llama[9:]
expl_factual_our_llama = factual_our_llama[:9]
info_factual_our_llama = factual_our_llama[9:]
expl_inclusion_our_llama = inclusion_our_llama[:9]
info_inclusion_our_llama = inclusion_our_llama[9:]
expl_exclusion_our_llama = exclusion_our_llama[:9]
info_exclusion_our_llama = exclusion_our_llama[9:]

scores = {}
scores['expl_fluent_our_human'] = expl_fluent_our_human
scores['info_fluent_our_human'] = info_fluent_our_human
scores['expl_factual_our_human'] = expl_factual_our_human
scores['info_factual_our_human'] = info_factual_our_human
scores['expl_inclusion_our_human'] = expl_inclusion_our_human
scores['info_inclusion_our_human'] = info_inclusion_our_human
scores['expl_exclusion_our_human'] = expl_exclusion_our_human
scores['info_exclusion_our_human'] = info_exclusion_our_human

scores['expl_fluent_our_gpt'] = expl_fluent_our_gpt
scores['info_fluent_our_gpt'] = info_fluent_our_gpt
scores['expl_factual_our_gpt'] = expl_factual_our_gpt
scores['info_factual_our_gpt'] = info_factual_our_gpt
scores['expl_inclusion_our_gpt'] = expl_inclusion_our_gpt
scores['info_inclusion_our_gpt'] = info_inclusion_our_gpt
scores['expl_exclusion_our_gpt'] = expl_exclusion_our_gpt
scores['info_exclusion_our_gpt'] = info_exclusion_our_gpt

scores['expl_fluent_our_llama'] = expl_fluent_our_llama
scores['info_fluent_our_llama'] = info_fluent_our_llama
scores['expl_factual_our_llama'] = expl_factual_our_llama
scores['info_factual_our_llama'] = info_factual_our_llama
scores['expl_inclusion_our_llama'] = expl_inclusion_our_llama
scores['info_inclusion_our_llama'] = info_inclusion_our_llama
scores['expl_exclusion_our_llama'] = expl_exclusion_our_llama
scores['info_exclusion_our_llama'] = info_exclusion_our_llama


```

# new times roman font

```python
import matplotlib.font_manager as fm
font_path = '/home/ywang1/data1/font/times new roman.ttf'
font_bold_path = '/home/ywang1/data1/font/times new roman bold.ttf'
times_new_roman = fm.FontProperties(fname=font_path)
times_new_roman_bold = fm.FontProperties(fname=font_bold_path)

metrics = ['Fluency', 'Factual Alignment', 'Inclusion', 'Exclusion']
intents = ['expl', 'info']
methods = ['human', 'gpt', 'llama']

expl_size = len(expl_fluent_our_human)
info_size = len(info_fluent_our_human)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
# custom_palette = sns.color_palette(["#EDC6B1", "#B7B7B7",  "#BFCCB5"])#"#7C96AB",
custom_palette = sns.color_palette(["#3D5C6F", "#E47159", "#F9AE78"])

for i, ax in enumerate(axes.flat):
    m = metrics[i]
    m = metrics[i] # start with fluency
    if m == "Fluency":
        metric = "fluent"
    elif m == "Factual Alignment":
        metric = "factual"
    elif m == "Inclusion":
        metric = "inclusion"
    elif m == "Exclusion":
        metric = "exclusion"
    data = {'intent': ['Exploratory'] * expl_size * 3 + ['Informative'] * info_size * 3, 
    'score': scores[f'expl_{metric}_our_human'] + scores[f'expl_{metric}_our_gpt'] + scores[f'expl_{metric}_our_llama'] + \
        scores[f'info_{metric}_our_human'] + scores[f'info_{metric}_our_gpt'] + scores[f'info_{metric}_our_llama'],
    'method': ['Human'] * expl_size + ['GPT-4o'] * expl_size + ['LLaMa3.1'] * expl_size + \
        ['Human'] * info_size + ['GPT-4o'] * info_size + ['LLaMa3.1'] * info_size}
    
    df = pd.DataFrame(data)
    sns.boxplot(x='intent', y='score', hue='method', data=df, ax=ax, palette=custom_palette, linewidth=3, width=0.7)
    ax.set_ylim(1, 5.2)
    sns.set(style="white", font=times_new_roman.get_name())
    # label
    ax.set_xticklabels(ax.get_xticklabels(), font=times_new_roman_bold, fontweight='bold')  # 首先设置所有标签以确保都被访问
    ax.set_yticklabels(ax.get_yticklabels(), font=times_new_roman_bold, fontweight='bold')
    ax.set_xlabel('Intent Types', font=times_new_roman_bold, fontweight='bold', fontsize=25)
    ax.set_ylabel('Scores', font=times_new_roman_bold, fontweight='bold', fontsize=25)

    # 增大轴刻度的字体大小
    ax.tick_params(axis='both', labelsize=20) # 对ax1的x轴和y轴刻度标签字体大小进行调整
    for label in ax.get_xticklabels():
        label.set_fontweight('bold')
    for label in ax.get_yticklabels():
        label.set_fontweight('bold')
    ax.set_title(m, font=times_new_roman_bold, fontweight='bold', fontsize=30)
    # matplotlib.rcParams.update({
    #     "font.family": 'Times New Roman',
    #     "font.size": 25,
    # })
    legend = ax.get_legend()  
    legend.set_title(None)
    # 设置边线宽度
    for spine in ax.spines.values():
        spine.set_linewidth(3)  # Set the spine line width to 2

    for text in legend.get_texts():
        text.set_fontproperties(times_new_roman_bold)
        text.set_fontweight('bold')
        # text.set_fontsize(20)
plt.tight_layout()
plt.savefig("evaluation_intent_type.pdf")

plt.show()
```
